```python
sigma0 = 5
mu = 1/sigma0
sample = np.random.exponential( 1/mu, int(1e6))

def funct(x, sigma, mu):
    return x**3 * np.exp(- (x**2 / (2*sigma**2))) / (mu * np.exp(- mu *x))


integral = np.mean(funct(sample, sigma0, mu))
print (integral, 2*sigma0**4)
```

```python
n = int(1e4)
sample = np.zeros(n)
numbers = int(1e6)
for i in range(int(1e6)):
    sample[i] = np.mean(funct(np.random.exponential(sigma0, numbers), sigma0, mu))
```
